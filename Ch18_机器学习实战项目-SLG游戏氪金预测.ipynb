```python
def paymentshow(x):
    alluser = len(x)
    paiduser = (x != 0).sum()
    totalpayment = x.sum()

    print(f'付费率: {paiduser/alluser*100:.2F}%')
    print(f'付费人数:{paiduser}')
    print(f'总收入:{totalpayment:.2f}')
    print(f'所有用户收入均值:{totalpayment/alluser:.2f}')
    print(f'付费用户收入均值:{totalpayment/paiduser:.2f}')
    print(f'前500名付费用户金额占比:{x.sort_values(ascending=False)[:500].sum()/totalpayment*100:.2f}%')
    print(f'前1000名付费用户金额占比:{x.sort_values(ascending=False)[:1000].sum()/totalpayment*100:.2f}%')
    print(f'前5000名付费用户金额占比:{x.sort_values(ascending=False)[:5000].sum()/totalpayment*100:.2f}%')

display(paymentshow(y7)) #七日的付费情况
display(paymentshow(y45)) #四十五日的付费情况
```

```python
#特征衍生
data['PVP胜率'] = data['PVP胜利次数']/data['PVP次数']
data['PVP主动发起率'] = data['主动发起PVP次数']/data['PVP次数']
```

```python
#分离资源类和等级类特征
grow = []
level = []
for a in data.columns:
    if '获取' in a:
        grow.append(a)
    elif '招募' in a:
        grow.append(a)
    elif '建筑' in a:
        level.append(a)
    elif '科研' in a:
        level.append(a)
    else:
        None

#特征衍生
data['玩家成长率'] = data.loc[:,grow].mean(axis=1) / data['在线时长']
data['付费成长率'] = data.loc[:,grow].mean(axis=1) / data['付费金额']
data['玩家升级率'] = data.loc[:,level].mean(axis=1) / data['在线时长']
data['付费升级率'] = data.loc[:,level].mean(axis=1) / data['付费金额']
```

```python
#缺失值填充为0
for a in ['PVP胜率', 'PVP主动发起率', '玩家成长率', '付费成长率', '玩家升级率', '付费升级率']:
    data.loc[data[a].isnull(), a]=0
    data.loc[data[a]==float('inf'), a]=0
```

```python
#七日内用户付费金额超过6标记为有潜力用户=1
data['潜力用户'] = (data['付费金额']>=6).astype(int)
#主动PVP发起率超过0.5标记为为好战用户=1
data['好战用户'] = (data['PVP主动发起率']>=0.05).astype(int)
#在线时长超过800标记为肝帝用户=0（应用匿名函数取反布尔值）
data['肝帝用户'] = (data['在线时长']>800).apply(lambda x: not x).astype(int)
#主动PVP发起率低于0.2或PVP次数=0的用户标记为佛系用户=0
data['佛系用户'] = ((data['PVP主动发起率']<0.2) | (data['PVP次数']==0)).apply(lambda x: not x).astype(int)
#PVP胜率小于0.1且PVP次数！=0的用户标记为菜鸟用户
data['菜鸟用户'] = ((data['PVP胜率']<0.1) & (data['PVP次数']!=0)).apply(lambda x: not x).astype(int)
#在线时长<15或基本资源木头获取数量<10000标记为流失用户
data['流失用户'] = ((data['在线时长']<15) | (data['木头获取数量']<10000)).apply(lambda x: not x).astype(int)
```

```python
#将7日无付费用户的异常值修改为中位数（7日付费用户的异常值不处理）
for i in Xtrain.columns:
    if i == '付费金额':
        continue
    if Xtrain[i].ndim > 1:
        # 强制转为 Series，如果是 DataFrame
        feature_train = Xtrain[i].iloc[:, 0]
        feature_test = Xtest[i].iloc[:, 0]
    else:
        feature_train = Xtrain[i]
        feature_test = Xtest[i]

    Q1 = np.quantile(feature_train, 0.25)
    Q3 = np.quantile(feature_train, 0.75)
    IQR = Q3 - Q1
    medium = np.median(feature_train)

    abnormal_train = (feature_train < (Q1 - 1.5 * IQR)) | (feature_train > (Q3 + 1.5 * IQR))
    abnormal_test = (feature_test < (Q1 - 1.5 * IQR)) | (feature_test > (Q3 + 1.5 * IQR))

    Xtrain.loc[(Xtrain['付费金额'] == 0) & abnormal_train, i] = medium
    Xtest.loc[(Xtest['付费金额'] == 0) & abnormal_test, i] = medium
```

```python
def optuna_objective(trial):
    
    n_estimators = trial.suggest_int("n_estimators",low=100,high=200,step=50)
    max_depth = trial.suggest_int("max_depth",low=2,high=4,step=1)
    max_features = trial.suggest_int("max_features",low=72,high=100)
    max_samples = trial.suggest_float("max_samples",low=0.5,high=1.0,step=0.1)
    min_samples_split = trial.suggest_int("min_samples_split",low=70,high=80)
    min_samples_leaf = trial.suggest_int("min_samples_leaf",low=3,high=8)

    
    model = RFR(n_estimators = n_estimators,
                max_depth = max_depth,
                max_features = max_features,
                max_samples = max_samples,
                min_samples_split = min_samples_split,
                min_samples_leaf = min_samples_leaf,
                bootstrap=True,
                oob_score=True,
                random_state=42,
                verbose=False,
                n_jobs=-1
                )
    
    KF = KFold(n_splits=5,shuffle=True,random_state=42)
    score = cross_validate(model,
						   Xtrain_LRC,
						   ytrain_LRC,
                           scoring="neg_root_mean_squared_error",
                           cv=KF,
                           verbose=False,
                           n_jobs=-1,
                           error_score='raise'
                           )
    
    return np.mean(np.abs(score["test_score"]))

def optimizer_optuna(n_trials): 

    algo = optuna.samplers.TPESampler(n_startup_trials = 10, n_ei_candidates = 24)

    opt = optuna.create_study(sampler = algo,
                              direction="minimize" 
                             )

    opt.optimize(optuna_objective,
                 n_trials=n_trials,
                 show_progress_bar=True
                )
    
    print(f'best_params:{opt.best_trial.params},best_score:{opt.best_trial.values}')
    
    return opt.best_trial.params, opt.best_trial.values
```
